```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('tajmahal.jpg')
```

```python
plt.imshow(image[:,:,::-1]) 
plt.title("Original Image")  
plt.axis('off')
plt.show()
```

```python
# Step 2: Image Translation
tx, ty = 100, 50  
M_translation = np.float32([[1, 0, tx], [0, 1, ty]])  
translated_image = cv2.warpAffine(image, M_translation, (237, 180))  
```

```python
image.shape
```

```python
plt.imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))  
plt.title("Translated Image")  
plt.axis('off')
plt.show()
```

```python
# Step 3: Image Scaling
fx, fy = 10.0, 5.0 
scaled_image = cv2.resize(image, None, fx=fx, fy=fy, interpolation=cv2.INTER_LINEAR)
```

```python
plt.imshow(cv2.cvtColor(scaled_image, cv2.COLOR_BGR2RGB))  # Display the scaled image
plt.title("Scaled Image")  # Set title
plt.axis('off')
plt.show()
```

```python
# Step 4: Image Shearing
shear_matrix = np.float32([[1, 0.5, 0], [0.5, 1, 0]]) 
sheared_image = cv2.warpAffine(image, shear_matrix, (237, 180))
```

```python
image.shape
```

```python
plt.imshow(cv2.cvtColor(sheared_image, cv2.COLOR_BGR2RGB))  
plt.title("Sheared Image")  # Set title
plt.axis('off')
```

```python
# Step 5: Image Reflection
reflected_image = cv2.flip(image, -2) 
```

```python
plt.imshow(cv2.cvtColor(reflected_image, cv2.COLOR_BGR2RGB))  
plt.title("Reflected Image")  # Set title
plt.axis('off')
plt.show()
```

```python
rotated_image = cv2.warpAffine(image, cv2.getRotationMatrix2D((115, 90), 45, 1), (230, 180))
```

```python
image.shape
```

```python
plt.imshow(rotated_image[:,:,::-1])  
plt.title("Rotated Image") 
plt.axis('off')
plt.show()
```

```python
# Step 7: Image Cropping
x, y, w, h = 100, 100, 200, 150 
cropped_image = image[y:y+h, x:x+w]
```

```python
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)) 
plt.title("Cropped Image") 
plt.axis('off')
```
